```python
# 🧩 Step 1: Import dependencies
import os
import requests
from dotenv import load_dotenv

# 🧰 Step 2: Load your Hugging Face API key
load_dotenv()  # make sure .env is in the same folder
HF_API_KEY = os.getenv("HUGGINGFACE_API_KEY")

# Check if the key loaded
if not HF_API_KEY:
    raise ValueError("⚠️ Hugging Face API key not found. Please add it to your .env file.")

# 🔗 Step 3: Define the model and API endpoint
MODEL_NAME = "facebook/bart-large-cnn"
API_URL = f"https://api-inference.huggingface.co/models/{MODEL_NAME}"
HEADERS = {"Authorization": f"Bearer {HF_API_KEY}"}

print("✅ Connected to Hugging Face model:", MODEL_NAME)
```

```python
# 📝 Step 4: Define the summarization function

def summarize_text(input_text: str, max_len: int = 150, min_len: int = 40):
    """
    Summarizes text using Hugging Face Inference API (no local model loading).
    """
    if not input_text.strip():
        return "⚠️ Please provide some text."

    payload = {
        "inputs": input_text[:2000],  # trim very long input
        "parameters": {"max_length": max_len, "min_length": min_len}
    }

    try:
        response = requests.post(API_URL, headers=HEADERS, json=payload, timeout=60)

        if response.status_code == 401:
            return "❌ Invalid API key. Please check your HUGGINGFACE_API_KEY in .env."
        elif response.status_code != 200:
            return f"❌ API error {response.status_code}: {response.text}"

        data = response.json()
        if isinstance(data, list) and "summary_text" in data[0]:
            return data[0]["summary_text"]
        elif isinstance(data, dict) and "error" in data:
            return f"⚠️ Model error: {data['error']}"
        else:
            return str(data)

    except Exception as e:
        return f"❌ Exception: {str(e)}"
```

```python
# 🧠 Step 5: Test the summarizer
sample_text = """
In 2024, artificial intelligence became an integral part of daily life,
powering everything from self-driving cars to personalized healthcare.
AI systems now assist doctors in diagnosis, optimize energy consumption in smart cities,
and even create art and music. However, concerns about bias, privacy, and ethics remain critical.
"""

summary = summarize_text(sample_text)
print("🧾 Summary:\n", summary)
```
